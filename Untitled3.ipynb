```python
# 1) IMPORT LIBRARIES
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error

# 2) DATA (Example — synthetic or placeholder)
# In practice, replace this with your CSV data loaded from a real source
data = {
    "Year": [2010,2012,2014,2016,2018,2020,2022,2024],
    "MedianPrice": [180000, 195000, 210000, 225000, 250000, 280000, 310000, 330000]
}
df = pd.DataFrame(data)

# 3) FEATURE ENGINEERING
df["YearNumeric"] = df["Year"] - df["Year"].min()

# 4) TRAIN-TEST SPLIT
X = df[["YearNumeric"]]
y = df["MedianPrice"]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, shuffle=False)

# 5) MODEL TRAINING
model = LinearRegression()
model.fit(X_train, y_train)

# 6) PREDICT 2026
year_2026 = pd.DataFrame({"YearNumeric": [2026 - df["Year"].min()]})
prediction_2026 = model.predict(year_2026)

print("Predicted Median Price for Summer 2026 (statewide estimate):")
print("$", round(prediction_2026[0], 2))

# 7) EVALUATE
y_pred_test = model.predict(X_test)
print("MAE on Test Set:", mean_absolute_error(y_test, y_pred_test))
```

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split, cross_val_score, GridSearchCV
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error
```

```python
data = {
    "Year": [2010,2012,2014,2016,2018,2020,2022,2024],
    "InterestRate": [4.7, 4.5, 4.2, 3.9, 4.6, 3.1, 5.5, 6.8],
    "PopulationGrowth": [1.6, 1.7, 1.8, 1.9, 2.0, 1.5, 1.4, 1.3],
    "MedianPrice": [180000,195000,210000,225000,250000,280000,310000,330000]
}

df = pd.DataFrame(data)
```

```python
# Feature engineering (ONLY ONCE)
df["YearIndex"] = df["Year"] - df["Year"].min()
```

```python
X = df[["YearIndex", "InterestRate", "PopulationGrowth"]]
y = df["MedianPrice"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, shuffle=False
)
```

```python
pipeline = Pipeline([
    ("scaler", StandardScaler()),
    ("model", RandomForestRegressor(
        n_estimators=300,
        random_state=42
    ))
])
```

```python
pipeline.fit(X_train, y_train)
```

```python
cv_scores = cross_val_score(
    pipeline, X_train, y_train,
    scoring="neg_mean_absolute_error",
    cv=3
)

print("Cross-validated MAE:", abs(cv_scores.mean()))
```

```python
future_2026 = pd.DataFrame({
    "YearIndex": [2026 - df["Year"].min()],
    "InterestRate": [5.5],
    "PopulationGrowth": [1.2]
})

prediction_2026 = pipeline.predict(future_2026)

print("Predicted Texas Median Home Price (Summer 2026):")
print("$", round(prediction_2026[0], 2))
```

```python
test_predictions = pipeline.predict(X_test)
mae = mean_absolute_error(y_test, test_predictions)

print("Test MAE:", round(mae, 2))
```
